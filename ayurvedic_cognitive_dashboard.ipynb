```python
# 🔧 Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.figure_factory as ff
import warnings
warnings.filterwarnings('ignore')

# Configure visualization settings
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
pd.set_option('display.max_columns', None)

print("🎨 Visualization libraries loaded successfully!")
print("📊 Dashboard ready for Ayurvedic-AI analysis...")
```

## 📂 Data Loading & Initial Exploration

```python
# 📊 Load Stage 1 - Prakriti Classification Dataset
try:
    df_prakriti = pd.read_csv('stage_1_prakriti_classification/data/Ayurveda_Prakriti_Dataset.csv')
    print(f"✅ Stage 1 Dataset Loaded: {df_prakriti.shape[0]:,} records, {df_prakriti.shape[1]} features")
    print(f"📊 Dosha Distribution: {df_prakriti['Dosha'].value_counts().to_dict()}")
except:
    print("⚠️ Stage 1 dataset not found. Please check file path.")

# 🧠 Load Stage 2 - Alzheimer's Risk Dataset
try:
    df_alzheimer = pd.read_csv('stage_2_alzheimer_risk_prediction/data/alzheimers_risk_dataset_stage2.csv')
    print(f"\n✅ Stage 2 Dataset Loaded: {df_alzheimer.shape[0]:,} records, {df_alzheimer.shape[1]} features")
    print(f"🎯 Risk Level Distribution: {df_alzheimer['risk_level'].value_counts().to_dict()}")
except:
    print("⚠️ Stage 2 dataset not found. Please check file path.")
```

## 🧬 Stage 1 Analysis: Prakriti Classification Insights

```python
# 🎨 Prakriti Distribution Visualization
if 'df_prakriti' in globals():
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=['Dosha Distribution', 'Body Frame by Dosha', 'Sleep Pattern Analysis', 'Energy Levels Distribution'],
        specs=[[{'type': 'domain'}, {'type': 'bar'}],
               [{'type': 'bar'}, {'type': 'violin'}]]
    )
    
    # Dosha pie chart
    dosha_counts = df_prakriti['Dosha'].value_counts()
    colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
    
    fig.add_trace(go.Pie(
        values=dosha_counts.values,
        labels=dosha_counts.index,
        marker_colors=colors,
        textinfo='label+percent',
        hovertemplate='<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>'
    ), row=1, col=1)
    
    # Body frame analysis
    body_dosha = pd.crosstab(df_prakriti['Body_Frame'], df_prakriti['Dosha'])
    for i, dosha in enumerate(body_dosha.columns):
        fig.add_trace(go.Bar(
            x=body_dosha.index,
            y=body_dosha[dosha],
            name=dosha,
            marker_color=colors[i]
        ), row=1, col=2)
    
    fig.update_layout(
        title_text="🧬 Ayurvedic Prakriti Analysis Dashboard",
        title_x=0.5,
        height=800,
        showlegend=True
    )
    
    fig.show()
else:
    print("📊 Prakriti dataset not available for visualization")
```

## 🧠 Stage 2 Analysis: Alzheimer's Risk Prediction Insights

```python
# 🎯 Alzheimer's Risk Analysis
if 'df_alzheimer' in globals():
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=['Risk Level Distribution', 'Risk by Prakriti Type', 'Age vs Risk Score', 'Symptoms Correlation'],
        specs=[[{'type': 'bar'}, {'type': 'box'}],
               [{'type': 'scatter'}, {'type': 'heatmap'}]]
    )
    
    # Risk level distribution
    risk_counts = df_alzheimer['risk_level'].value_counts()
    risk_colors = {'Low': '#2ECC71', 'Medium': '#F39C12', 'High': '#E74C3C'}
    
    fig.add_trace(go.Bar(
        x=risk_counts.index,
        y=risk_counts.values,
        marker_color=[risk_colors.get(x, '#BDC3C7') for x in risk_counts.index],
        text=risk_counts.values,
        textposition='auto'
    ), row=1, col=1)
    
    # Risk by Prakriti
    for prakriti in df_alzheimer['prakriti_type'].unique():
        prakriti_data = df_alzheimer[df_alzheimer['prakriti_type'] == prakriti]['risk_score']
        fig.add_trace(go.Box(
            y=prakriti_data,
            name=prakriti,
            boxpoints='outliers'
        ), row=1, col=2)
    
    # Age vs Risk Score scatter
    fig.add_trace(go.Scatter(
        x=df_alzheimer['age'],
        y=df_alzheimer['risk_score'],
        mode='markers',
        marker=dict(
            color=df_alzheimer['risk_score'],
            colorscale='RdYlBu_r',
            size=8,
            colorbar=dict(title="Risk Score")
        ),
        text=df_alzheimer['prakriti_type'],
        hovertemplate='Age: %{x}<br>Risk Score: %{y}<br>Prakriti: %{text}<extra></extra>'
    ), row=2, col=1)
    
    fig.update_layout(
        title_text="🧠 Alzheimer's Risk Prediction Analysis Dashboard",
        title_x=0.5,
        height=800,
        showlegend=False
    )
    
    fig.show()
    
    # Summary statistics
    print("\n📊 Key Insights:")
    print(f"👥 Total Patients Analyzed: {len(df_alzheimer):,}")
    print(f"⚠️ High Risk Patients: {len(df_alzheimer[df_alzheimer['risk_level'] == 'High']):,}")
    print(f"📈 Average Risk Score: {df_alzheimer['risk_score'].mean():.1f}")
    print(f"👴 Average Age: {df_alzheimer['age'].mean():.1f} years")
else:
    print("📊 Alzheimer dataset not available for visualization")
```

## 🔬 Model Performance Analysis

```python
# 📈 Model Performance Metrics Simulation
# (In production, load actual model metrics from saved files)

performance_data = {
    'Model': ['Prakriti Classifier', 'Alzheimer Risk Predictor'],
    'Accuracy': [0.94, 0.87],
    'Precision': [0.93, 0.89],
    'Recall': [0.92, 0.85],
    'F1-Score': [0.925, 0.87]
}

df_performance = pd.DataFrame(performance_data)

# Create radar chart for model comparison
categories = ['Accuracy', 'Precision', 'Recall', 'F1-Score']

fig = go.Figure()

for idx, model in enumerate(df_performance['Model']):
    values = df_performance.iloc[idx][1:].tolist()
    values.append(values[0])  # Complete the circle
    
    fig.add_trace(go.Scatterpolar(
        r=values,
        theta=categories + [categories[0]],
        fill='toself',
        name=model,
        line_color=['#FF6B6B', '#4ECDC4'][idx]
    ))

fig.update_layout(
    polar=dict(
        radialaxis=dict(visible=True, range=[0.8, 1.0])
    ),
    title="🎯 Model Performance Comparison",
    title_x=0.5
)

fig.show()

print("\n🏆 Model Performance Summary:")
print(df_performance.to_string(index=False))
```

## 💊 Ayurvedic Recommendations Analysis

```python
# 🌿 Ayurvedic Recommendations Distribution
if 'df_alzheimer' in globals():
    # Extract common Ayurvedic herbs mentioned in recommendations
    ayurvedic_herbs = {
        'Brahmi': 'Memory enhancement',
        'Ashwagandha': 'Stress reduction', 
        'Shankhpushpi': 'Cognitive improvement',
        'Gotu Kola': 'Brain function',
        'Triphala': 'Detoxification',
        'Guggulu': 'Anti-inflammatory'
    }
    
    # Count herb mentions in recommendations
    herb_counts = {}
    for herb in ayurvedic_herbs:
        if 'ayurveda_recommendations' in df_alzheimer.columns:
            count = df_alzheimer['ayurveda_recommendations'].str.contains(herb, na=False).sum()
            herb_counts[herb] = count
    
    if herb_counts:
        fig = go.Figure(data=[go.Bar(
            x=list(herb_counts.keys()),
            y=list(herb_counts.values()),
            marker_color='#2ECC71',
            text=list(herb_counts.values()),
            textposition='auto'
        )])
        
        fig.update_layout(
            title="🌿 Most Recommended Ayurvedic Herbs",
            xaxis_title="Herbs",
            yaxis_title="Recommendation Frequency",
            title_x=0.5
        )
        
        fig.show()
    
    print("\n🌿 Ayurvedic Herb Benefits:")
    for herb, benefit in ayurvedic_herbs.items():
        print(f"• {herb}: {benefit}")
else:
    print("📊 Recommendation data not available")
```
