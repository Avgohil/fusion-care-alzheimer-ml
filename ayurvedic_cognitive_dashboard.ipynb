## 🔄 Dynamic Interactive Controls

**Real-time assessment input - changes update charts instantly!**

```python
# Create interactive widgets for real-time assessment input
class DynamicAssessmentDashboard:
    def __init__(self):
        self.output = widgets.Output()
        self.create_widgets()
        
    def create_widgets(self):
        """Create all interactive input widgets"""
        
        # User Info
        self.user_id = widgets.Text(value='USER_001', description='User ID:')
        
        # Ayurvedic Constitution Sliders (0-100, higher = more imbalanced)
        self.vata_slider = widgets.IntSlider(value=75, min=0, max=100, description='Vata Score:', style={'description_width': 'initial'})
        self.pitta_slider = widgets.IntSlider(value=45, min=0, max=100, description='Pitta Score:', style={'description_width': 'initial'})
        self.kapha_slider = widgets.IntSlider(value=30, min=0, max=100, description='Kapha Score:', style={'description_width': 'initial'})
        
        # Physical Health Sliders (0-100, higher = better)
        self.sleep_slider = widgets.IntSlider(value=35, min=0, max=100, description='Sleep Quality:', style={'description_width': 'initial'})
        self.digestion_slider = widgets.IntSlider(value=60, min=0, max=100, description='Digestion Score:', style={'description_width': 'initial'})
        self.energy_slider = widgets.IntSlider(value=40, min=0, max=100, description='Energy Levels:', style={'description_width': 'initial'})
        self.stress_slider = widgets.IntSlider(value=80, min=0, max=100, description='Stress Level:', style={'description_width': 'initial'})
        
        # Cognitive Health Sliders (0-100, higher = better)
        self.memory_slider = widgets.IntSlider(value=70, min=0, max=100, description='Memory Score:', style={'description_width': 'initial'})
        self.attention_slider = widgets.IntSlider(value=45, min=0, max=100, description='Attention Score:', style={'description_width': 'initial'})
        self.processing_slider = widgets.IntSlider(value=55, min=0, max=100, description='Processing Speed:', style={'description_width': 'initial'})
        self.executive_slider = widgets.IntSlider(value=40, min=0, max=100, description='Executive Function:', style={'description_width': 'initial'})
        
        # Control buttons
        self.auto_update = widgets.Checkbox(value=True, description='Auto Update Charts')
        self.random_data_btn = widgets.Button(description='🎲 Generate Random Data', button_style='info')
        self.reset_btn = widgets.Button(description='🔄 Reset to Default', button_style='warning')
        
        # Bind events
        self.setup_observers()
        
    def setup_observers(self):
        """Setup real-time observers for all widgets"""
        widgets_list = [
            self.vata_slider, self.pitta_slider, self.kapha_slider,
            self.sleep_slider, self.digestion_slider, self.energy_slider, self.stress_slider,
            self.memory_slider, self.attention_slider, self.processing_slider, self.executive_slider
        ]
        
        for widget in widgets_list:
            widget.observe(self.on_value_change, names='value')
            
        self.random_data_btn.on_click(self.generate_random_data)
        self.reset_btn.on_click(self.reset_to_default)
        
    def get_current_assessment_data(self):
        """Get current assessment data from widgets"""
        
        # Calculate overall risk score based on current inputs
        constitution_avg = np.mean([self.vata_slider.value, self.pitta_slider.value, self.kapha_slider.value])
        physical_avg = np.mean([self.sleep_slider.value, self.digestion_slider.value, self.energy_slider.value, 100-self.stress_slider.value])
        cognitive_avg = np.mean([self.memory_slider.value, self.attention_slider.value, self.processing_slider.value, self.executive_slider.value])
        
        # Weight the overall risk (constitution imbalance, physical health, cognitive health)
        overall_risk = (constitution_avg * 0.3) + ((100-physical_avg) * 0.4) + ((100-cognitive_avg) * 0.3)
        
        # Determine risk category
        if overall_risk <= 30:
            risk_category = 'Low Risk'
        elif overall_risk <= 60:
            risk_category = 'Moderate Risk'
        else:
            risk_category = 'High Risk'
            
        # Determine dominant constitution
        constitution_scores = [self.vata_slider.value, self.pitta_slider.value, self.kapha_slider.value]
        dominant_constitution = ['Vata', 'Pitta', 'Kapha'][np.argmax(constitution_scores)]
        
        return {
            'user_id': self.user_id.value,
            'assessment_date': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'vata_score': self.vata_slider.value,
            'pitta_score': self.pitta_slider.value,
            'kapha_score': self.kapha_slider.value,
            'sleep_quality': self.sleep_slider.value,
            'digestion_score': self.digestion_slider.value,
            'energy_levels': self.energy_slider.value,
            'stress_level': self.stress_slider.value,
            'memory_score': self.memory_slider.value,
            'attention_score': self.attention_slider.value,
            'processing_speed': self.processing_slider.value,
            'executive_function': self.executive_slider.value,
            'overall_risk_score': int(overall_risk),
            'risk_category': risk_category,
            'dominant_constitution': dominant_constitution
        }
        
    def on_value_change(self, change):
        """Handle real-time updates when slider values change"""
        if self.auto_update.value:
            self.update_charts()
            
    def generate_random_data(self, btn):
        """Generate random assessment data"""
        self.vata_slider.value = random.randint(20, 90)
        self.pitta_slider.value = random.randint(20, 90)
        self.kapha_slider.value = random.randint(20, 90)
        self.sleep_slider.value = random.randint(20, 90)
        self.digestion_slider.value = random.randint(30, 90)
        self.energy_slider.value = random.randint(25, 85)
        self.stress_slider.value = random.randint(30, 90)
        self.memory_slider.value = random.randint(35, 95)
        self.attention_slider.value = random.randint(30, 90)
        self.processing_slider.value = random.randint(35, 85)
        self.executive_slider.value = random.randint(30, 85)
        
    def reset_to_default(self, btn):
        """Reset all sliders to default values"""
        self.vata_slider.value = 75
        self.pitta_slider.value = 45
        self.kapha_slider.value = 30
        self.sleep_slider.value = 35
        self.digestion_slider.value = 60
        self.energy_slider.value = 40
        self.stress_slider.value = 80
        self.memory_slider.value = 70
        self.attention_slider.value = 45
        self.processing_slider.value = 55
        self.executive_slider.value = 40
        
    def update_charts(self):
        """Update all charts with current data"""
        current_data = self.get_current_assessment_data()
        
        with self.output:
            clear_output(wait=True)
            
            # Create all charts with current data
            risk_gauge = create_risk_gauge(current_data)
            constitution_chart = create_constitution_chart(current_data)
            health_radar = create_health_radar(current_data)
            cognitive_summary = create_cognitive_summary(current_data)
            compact_summary = create_compact_result_summary(current_data)
            
            print(f"🔄 Charts updated at: {current_data['assessment_date']}")
            print(f"📊 Current Risk Level: {current_data['risk_category']} ({current_data['overall_risk_score']}%)")
            print(f"🏛️ Dominant Constitution: {current_data['dominant_constitution']}")
            print("-" * 60)
            
            # Display charts
            risk_gauge.show()
            constitution_chart.show()
            health_radar.show()
            cognitive_summary.show()
            compact_summary.show()
            
    def display_dashboard(self):
        """Display the complete interactive dashboard"""
        
        # Create layout
        constitution_box = widgets.VBox([
            widgets.HTML("<h3>🏛️ Ayurvedic Constitution (0-100, higher = more imbalanced)</h3>"),
            self.vata_slider, self.pitta_slider, self.kapha_slider
        ])
        
        physical_box = widgets.VBox([
            widgets.HTML("<h3>💪 Physical Health (0-100, higher = better)</h3>"),
            self.sleep_slider, self.digestion_slider, self.energy_slider,
            widgets.HTML("<p><i>Stress Level (0-100, higher = more stress):</i></p>"),
            self.stress_slider
        ])
        
        cognitive_box = widgets.VBox([
            widgets.HTML("<h3>🧠 Cognitive Health (0-100, higher = better)</h3>"),
            self.memory_slider, self.attention_slider, self.processing_slider, self.executive_slider
        ])
        
        controls_box = widgets.VBox([
            widgets.HTML("<h3>⚙️ Controls</h3>"),
            self.user_id, self.auto_update, self.random_data_btn, self.reset_btn
        ])
        
        # Layout in rows
        input_layout = widgets.HBox([constitution_box, physical_box, cognitive_box, controls_box])
        
        # Complete dashboard
        dashboard = widgets.VBox([
            widgets.HTML("<h1>🔄 DYNAMIC Assessment Dashboard</h1>"),
            widgets.HTML("<p><i>Adjust sliders below to see charts update in real-time!</i></p>"),
            input_layout,
            widgets.HTML("<hr>"),
            widgets.HTML("<h2>📊 Live Chart Updates:</h2>"),
            self.output
        ])
        
        display(dashboard)
        
        # Initial chart update
        self.update_charts()

# Create the dynamic dashboard instance
print("🎯 Creating dynamic assessment dashboard...")
dynamic_dashboard = DynamicAssessmentDashboard()
print("✅ Dynamic dashboard ready! Use .display_dashboard() to show interactive controls.")
```

```python
# Risk color coding function
def get_risk_color(score, parameter_type='standard'):
    """Return color based on risk level"""
    if parameter_type == 'cognitive' or parameter_type == 'physical':
        # Higher is better for these parameters
        if score >= 70: return '#2ecc71'  # Green - Good
        elif score >= 40: return '#f39c12'  # Orange - Moderate
        else: return '#e74c3c'  # Red - Poor
    elif parameter_type == 'constitution':
        # For constitution, balanced (lower) is better
        if score <= 40: return '#2ecc71'  # Green - Balanced
        elif score <= 65: return '#f39c12'  # Orange - Moderate imbalance
        else: return '#e74c3c'  # Red - High imbalance
    elif parameter_type == 'stress':
        # For stress, lower is better
        if score <= 30: return '#2ecc71'  # Green - Low stress
        elif score <= 60: return '#f39c12'  # Orange - Moderate stress
        else: return '#e74c3c'  # Red - High stress

# Color palette
colors = {
    'success': '#2ecc71',
    'warning': '#f39c12', 
    'danger': '#e74c3c',
    'primary': '#3498db',
    'dark': '#2c3e50'
}

print("🎨 Risk color coding system ready!")
print("   🟢 Green: Good/Balanced")
print("   🟡 Orange: Moderate Risk") 
print("   🔴 Red: High Risk")
```

## 📊 1. Overall Risk Level Gauge

**Simple, prominent risk indicator - perfect for form results**

```python
def create_risk_gauge(assessment_data):
    """Create a clean risk level gauge - perfect for form results"""
    
    risk_score = assessment_data['overall_risk_score']
    risk_category = assessment_data['risk_category']
    
    # Determine gauge color
    if risk_score <= 30:
        gauge_color = colors['success']
        status_text = "Low Risk ✅"
    elif risk_score <= 60:
        gauge_color = colors['warning']
        status_text = "Moderate Risk ⚠️"
    else:
        gauge_color = colors['danger']
        status_text = "High Risk ❗"
    
    # Create simple gauge
    fig = go.Figure(go.Indicator(
        mode = "gauge+number",
        value = risk_score,
        title = {'text': f"Overall Health Risk Level"},
        gauge = {
            'axis': {'range': [0, 100], 'tickwidth': 2},
            'bar': {'color': gauge_color, 'thickness': 0.3},
            'bgcolor': "white",
            'borderwidth': 2,
            'bordercolor': "lightgray",
            'steps': [
                {'range': [0, 30], 'color': 'rgba(46, 204, 113, 0.2)'},
                {'range': [30, 60], 'color': 'rgba(243, 156, 18, 0.2)'},
                {'range': [60, 100], 'color': 'rgba(231, 76, 60, 0.2)'}
            ],
        }
    ))
    
    # Add status text
    fig.add_annotation(
        text=f"<b>{status_text}</b>",
        x=0.5, y=0.15,
        font=dict(size=18, color=gauge_color),
        showarrow=False
    )
    
    fig.update_layout(
        height=300,
        font={'family': "Arial", 'size': 14},
        margin=dict(l=50, r=50, t=50, b=50),
        paper_bgcolor='white'
    )
    
    return fig

# Create and display risk gauge
risk_gauge = create_risk_gauge(user_assessment)
risk_gauge.show()

print("🎯 Risk gauge created - perfect for displaying after form submission!")
```

## 🏛️ 2. Constitution Balance Chart

**Simple bar chart showing Ayurvedic constitution balance**

```python
def create_constitution_chart(assessment_data):
    """Create clean constitution balance chart"""
    
    # Constitution data
    constitutions = ['Vata', 'Pitta', 'Kapha']
    scores = [
        assessment_data['vata_score'], 
        assessment_data['pitta_score'], 
        assessment_data['kapha_score']
    ]
    
    # Get colors based on balance level
    bar_colors = [get_risk_color(score, 'constitution') for score in scores]
    
    # Create bar chart
    fig = go.Figure(data=[
        go.Bar(
            x=constitutions,
            y=scores,
            marker_color=bar_colors,
            text=[f'{score}%' for score in scores],
            textposition='auto',
            hovertemplate='<b>%{x} Constitution</b><br>' +
                         'Imbalance Level: %{y}%<br>' +
                         '<i>Lower is better (more balanced)</i><extra></extra>'
        )
    ])
    
    fig.update_layout(
        title=dict(
            text="Ayurvedic Constitution Balance",
            x=0.5,
            font=dict(size=16)
        ),
        xaxis_title="Constitution Type",
        yaxis_title="Imbalance Level (%)",
        yaxis_range=[0, 100],
        height=350,
        font=dict(family="Arial", size=12),
        margin=dict(l=50, r=50, t=80, b=50),
        paper_bgcolor='white'
    )
    
    # Add dominant constitution annotation
    dominant_idx = scores.index(max(scores))
    dominant_constitution = constitutions[dominant_idx]
    
    fig.add_annotation(
        text=f"Dominant: <b>{dominant_constitution}</b>",
        x=dominant_idx, y=scores[dominant_idx] + 5,
        font=dict(size=14, color=bar_colors[dominant_idx]),
        showarrow=True,
        arrowhead=2,
        arrowcolor=bar_colors[dominant_idx]
    )
    
    return fig

# Create and display constitution chart
constitution_chart = create_constitution_chart(user_assessment)
constitution_chart.show()

print("🏛️ Constitution balance chart created!")
```

## 🕸️ 3. Health Overview Radar Chart

**Compact radar chart showing all health parameters at once**

```python
def create_health_radar(assessment_data):
    """Create compact radar chart for health overview"""
    
    # Parameters for radar chart
    parameters = [
        'Sleep Quality', 'Digestion', 'Energy Levels', 'Stress Management',
        'Memory', 'Attention', 'Processing Speed', 'Executive Function'
    ]
    
    # Normalize scores (invert stress for better visualization)
    scores = [
        assessment_data['sleep_quality'],
        assessment_data['digestion_score'],
        assessment_data['energy_levels'],
        100 - assessment_data['stress_level'],  # Invert stress
        assessment_data['memory_score'],
        assessment_data['attention_score'],
        assessment_data['processing_speed'],
        assessment_data['executive_function']
    ]
    
    # Determine overall color based on average score
    avg_score = np.mean(scores)
    radar_color = get_risk_color(avg_score, 'physical')
    
    # Convert hex color to RGB for transparency
    def hex_to_rgba(hex_color, alpha=0.3):
        """Convert hex color to rgba format"""
        hex_color = hex_color.lstrip('#')
        r = int(hex_color[0:2], 16)
        g = int(hex_color[2:4], 16)
        b = int(hex_color[4:6], 16)
        return f'rgba({r},{g},{b},{alpha})'
    
    fig = go.Figure()
    
    # Add user's scores
    fig.add_trace(go.Scatterpolar(
        r=scores,
        theta=parameters,
        fill='toself',
        fillcolor=hex_to_rgba(radar_color, 0.3),  # Use proper rgba format
        line=dict(color=radar_color, width=3),
        marker=dict(size=6, color=radar_color),
        name='Your Results',
        hovertemplate='<b>%{theta}</b><br>Score: %{r}%<extra></extra>'
    ))
    
    # Add target/ideal line
    fig.add_trace(go.Scatterpolar(
        r=[80] * len(parameters),
        theta=parameters,
        fill='toself',
        fillcolor='rgba(46, 204, 113, 0.1)',
        line=dict(color='rgba(46, 204, 113, 0.6)', width=2, dash='dash'),
        name='Target Level',
        hovertemplate='Target: %{r}%<extra></extra>'
    ))
    
    fig.update_layout(
        polar=dict(
            radialaxis=dict(
                visible=True,
                range=[0, 100],
                tickfont=dict(size=10),
                gridcolor='rgba(0,0,0,0.1)'
            ),
            angularaxis=dict(
                tickfont=dict(size=11),
                rotation=90
            )
        ),
        title=dict(
            text="Health Overview - All Parameters",
            x=0.5,
            font=dict(size=16)
        ),
        height=400,
        showlegend=True,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="center",
            x=0.5
        ),
        margin=dict(l=50, r=50, t=80, b=50),
        paper_bgcolor='white'
    )
    
    return fig

# Create and display health radar
health_radar = create_health_radar(user_assessment)
health_radar.show()

print("🕸️ Health overview radar chart created!")
```

## 🧠 4. Cognitive Health Summary

**Simple bar chart for cognitive assessment results**

```python
def create_cognitive_summary(assessment_data):
    """Create clean cognitive health summary chart"""
    
    # Cognitive parameters
    cognitive_params = ['Memory', 'Attention', 'Processing Speed', 'Executive Function']
    cognitive_scores = [
        assessment_data['memory_score'],
        assessment_data['attention_score'], 
        assessment_data['processing_speed'],
        assessment_data['executive_function']
    ]
    
    # Get colors for each cognitive parameter
    cognitive_colors = [get_risk_color(score, 'cognitive') for score in cognitive_scores]
    
    # Create horizontal bar chart for better readability
    fig = go.Figure(data=[
        go.Bar(
            y=cognitive_params,
            x=cognitive_scores,
            orientation='h',
            marker_color=cognitive_colors,
            text=[f'{score}%' for score in cognitive_scores],
            textposition='auto',
            hovertemplate='<b>%{y}</b><br>' +
                         'Score: %{x}%<br>' +
                         '<i>Higher is better</i><extra></extra>'
        )
    ])
    
    fig.update_layout(
        title=dict(
            text="Cognitive Health Assessment",
            x=0.5,
            font=dict(size=16)
        ),
        xaxis_title="Score (%)",
        xaxis_range=[0, 100],
        height=300,
        font=dict(family="Arial", size=12),
        margin=dict(l=120, r=50, t=80, b=50),
        paper_bgcolor='white'
    )
    
    # Add average line
    avg_cognitive = np.mean(cognitive_scores)
    fig.add_vline(
        x=avg_cognitive, 
        line_dash="dash", 
        line_color=get_risk_color(avg_cognitive, 'cognitive'),
        annotation_text=f"Average: {avg_cognitive:.0f}%",
        annotation_position="top"
    )
    
    return fig

# Create and display cognitive summary
cognitive_summary = create_cognitive_summary(user_assessment)
cognitive_summary.show()

print("🧠 Cognitive health summary created!")
```

## 🌐 5. Web Integration Functions

**Ready-to-use functions for your form workflow**

```python
# 🛡️ DATA VALIDATION & ERROR HANDLING
# Enhanced validation for dynamic data inputs

class DataValidator:
    @staticmethod
    def validate_assessment_data(data):
        """Validate assessment data structure and values"""
        required_fields = [
            'user_id', 'vata_score', 'pitta_score', 'kapha_score',
            'sleep_quality', 'digestion_score', 'energy_levels', 'stress_level',
            'memory_score', 'attention_score', 'processing_speed', 'executive_function'
        ]
        
        errors = []
        warnings = []
        
        # Check required fields
        for field in required_fields:
            if field not in data:
                errors.append(f"Missing required field: {field}")
                
        # Validate score ranges (0-100)
        score_fields = [f for f in required_fields if f != 'user_id']
        for field in score_fields:
            if field in data:
                value = data[field]
                if not isinstance(value, (int, float)):
                    errors.append(f"{field} must be a number")
                elif value < 0 or value > 100:
                    errors.append(f"{field} must be between 0 and 100 (got {value})")
                elif value < 10:
                    warnings.append(f"{field} is very low ({value}%) - please verify")
                elif value > 90:
                    warnings.append(f"{field} is very high ({value}%) - please verify")
                    
        # Validate user_id
        if 'user_id' in data:
            if not isinstance(data['user_id'], str) or len(data['user_id']) == 0:
                errors.append("user_id must be a non-empty string")
                
        # Constitution balance check
        if all(f in data for f in ['vata_score', 'pitta_score', 'kapha_score']):
            constitution_sum = data['vata_score'] + data['pitta_score'] + data['kapha_score']
            if constitution_sum > 240:  # Very high total imbalance
                warnings.append(f"Total constitution imbalance is very high ({constitution_sum}) - consider reviewing")
                
        return {
            'valid': len(errors) == 0,
            'errors': errors,
            'warnings': warnings
        }
        
    @staticmethod
    def sanitize_data(data):
        """Clean and sanitize input data"""
        sanitized = data.copy()
        
        # Ensure numeric fields are integers
        numeric_fields = ['vata_score', 'pitta_score', 'kapha_score', 'sleep_quality', 
                         'digestion_score', 'energy_levels', 'stress_level', 'memory_score', 
                         'attention_score', 'processing_speed', 'executive_function']
        
        for field in numeric_fields:
            if field in sanitized:
                try:
                    # Convert to int and clamp to 0-100 range
                    sanitized[field] = max(0, min(100, int(float(sanitized[field]))))
                except (ValueError, TypeError):
                    sanitized[field] = 50  # Default to middle value
                    
        # Clean user_id
        if 'user_id' in sanitized:
            sanitized['user_id'] = str(sanitized['user_id']).strip()[:50]  # Limit length
            
        return sanitized

# Enhanced dashboard with validation
class ValidatedDynamicDashboard(DynamicAssessmentDashboard):
    def __init__(self):
        super().__init__()
        self.validator = DataValidator()
        self.validation_output = widgets.Output()
        
    def get_current_assessment_data(self):
        """Get validated assessment data"""
        raw_data = super().get_current_assessment_data()
        
        # Sanitize data
        clean_data = self.validator.sanitize_data(raw_data)
        
        # Validate data
        validation_result = self.validator.validate_assessment_data(clean_data)
        
        # Show validation results
        with self.validation_output:
            clear_output(wait=True)
            if validation_result['valid']:
                print("✅ Data validation passed")
            else:
                print("❌ Data validation failed:")
                for error in validation_result['errors']:
                    print(f"   • {error}")
                    
            if validation_result['warnings']:
                print("⚠️ Validation warnings:")
                for warning in validation_result['warnings']:
                    print(f"   • {warning}")
        
        return clean_data if validation_result['valid'] else None
        
    def update_charts(self):
        """Update charts with validated data"""
        current_data = self.get_current_assessment_data()
        
        if current_data is None:
            with self.output:
                clear_output(wait=True)
                print("❌ Cannot update charts: Data validation failed")
            return
            
        # Call parent update method
        super().update_charts()
        
    def display_dashboard(self):
        """Display dashboard with validation panel"""
        # Get original dashboard
        super().display_dashboard()
        
        # Add validation panel
        validation_panel = widgets.VBox([
            widgets.HTML("<h3>🛡️ Data Validation Status</h3>"),
            self.validation_output
        ])
        
        display(validation_panel)

# Error handling decorator
def handle_chart_errors(func):
    """Decorator to handle chart creation errors gracefully"""
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            print(f"⚠️ Chart creation error in {func.__name__}: {str(e)}")
            print("🔄 Falling back to basic chart...")
            
            # Create a simple error chart
            import plotly.graph_objects as go
            fig = go.Figure()
            fig.add_annotation(
                text=f"Chart Error: {str(e)[:100]}...",
                x=0.5, y=0.5,
                font=dict(size=16, color="red"),
                showarrow=False
            )
            fig.update_layout(
                title="Chart Creation Error",
                height=300,
                showlegend=False
            )
            return fig
    return wrapper

# Apply error handling to all chart functions
create_risk_gauge = handle_chart_errors(create_risk_gauge)
create_constitution_chart = handle_chart_errors(create_constitution_chart)
create_health_radar = handle_chart_errors(create_health_radar)
create_cognitive_summary = handle_chart_errors(create_cognitive_summary)
create_compact_result_summary = handle_chart_errors(create_compact_result_summary)

print("🛡️ Data validation and error handling enabled!")
print("✅ All chart functions now have error handling")
print("🔍 Input validation active for all dynamic data")

# Create validated dashboard
print("\n🚀 Creating FULLY DYNAMIC dashboard with validation...")
validated_dashboard = ValidatedDynamicDashboard()
print("✅ Enhanced dynamic dashboard ready!")
```

```python
# 🌐 WEB INTEGRATION - Convert Dynamic Dashboard to Web App
# Complete FastAPI integration with real-time WebSocket updates

from fastapi import FastAPI, WebSocket, WebSocketDisconnect, HTTPException
from fastapi.responses import HTMLResponse, JSONResponse
from fastapi.staticfiles import StaticFiles
from fastapi.middleware.cors import CORSMiddleware
import plotly.io as pio
import plotly.graph_objects as go
import json
import asyncio
from typing import List, Dict, Any
import uvicorn
from datetime import datetime
import threading

class WebDashboardManager:
    def __init__(self):
        self.app = FastAPI(title="Dynamic Assessment Dashboard API", version="1.0.0")
        self.active_connections: List[WebSocket] = []
        self.data_manager = DynamicDataManager()
        self.validator = DataValidator()
        self.setup_routes()
        self.setup_middleware()
        
    def setup_middleware(self):
        """Setup CORS and other middleware"""
        self.app.add_middleware(
            CORSMiddleware,
            allow_origins=["*"],  # Configure for production
            allow_credentials=True,
            allow_methods=["*"],
            allow_headers=["*"],
        )
        
    def setup_routes(self):
        """Setup all API routes"""
        
        @self.app.get("/", response_class=HTMLResponse)
        async def get_dashboard():
            """Serve the main dashboard HTML"""
            return self.get_dashboard_html()
            
        @self.app.post("/api/assessment/create")
        async def create_assessment(assessment_data: dict):
            """Create new assessment with validation"""
            try:
                # Validate data
                validation_result = self.validator.validate_assessment_data(assessment_data)
                if not validation_result['valid']:
                    raise HTTPException(status_code=400, detail={
                        "errors": validation_result['errors'],
                        "warnings": validation_result['warnings']
                    })
                
                # Sanitize data
                clean_data = self.validator.sanitize_data(assessment_data)
                
                # Generate charts
                charts = self.generate_dynamic_charts(clean_data)
                
                # Broadcast to all connected clients
                await self.broadcast_update(clean_data, charts)
                
                return {
                    "status": "success",
                    "data": clean_data,
                    "charts": charts,
                    "validation": validation_result
                }
                
            except Exception as e:
                raise HTTPException(status_code=500, detail=f"Error creating assessment: {str(e)}")
                
        @self.app.get("/api/users")
        async def get_users():
            """Get all available users"""
            return {"users": self.data_manager.get_all_users()}
            
        @self.app.get("/api/users/{user_id}")
        async def get_user_data(user_id: str):
            """Get specific user's assessment data"""
            user_data = self.data_manager.get_user_data(user_id)
            if not user_data:
                raise HTTPException(status_code=404, detail="User not found")
            return user_data
            
        @self.app.post("/api/charts/generate")
        async def generate_charts_api(assessment_data: dict):
            """Generate charts for given assessment data"""
            try:
                charts = self.generate_dynamic_charts(assessment_data)
                return {"charts": charts}
            except Exception as e:
                raise HTTPException(status_code=500, detail=f"Error generating charts: {str(e)}")
                
        @self.app.websocket("/ws")
        async def websocket_endpoint(websocket: WebSocket):
            """WebSocket for real-time updates"""
            await self.connect_websocket(websocket)
            
    async def connect_websocket(self, websocket: WebSocket):
        """Handle WebSocket connections"""
        await websocket.accept()
        self.active_connections.append(websocket)
        
        try:
            while True:
                # Keep connection alive and handle messages
                data = await websocket.receive_text()
                message = json.loads(data)
                
                if message.get("type") == "request_update":
                    user_id = message.get("user_id", "USER_001")
                    user_data = self.data_manager.get_user_data(user_id)
                    if user_data:
                        charts = self.generate_dynamic_charts(user_data)
                        await websocket.send_text(json.dumps({
                            "type": "chart_update",
                            "data": user_data,
                            "charts": charts
                        }))
                        
        except WebSocketDisconnect:
            self.active_connections.remove(websocket)
            
    async def broadcast_update(self, data: dict, charts: dict):
        """Broadcast updates to all connected clients"""
        message = json.dumps({
            "type": "broadcast_update",
            "data": data,
            "charts": charts,
            "timestamp": datetime.now().isoformat()
        })
        
        disconnected = []
        for connection in self.active_connections:
            try:
                await connection.send_text(message)
            except Exception:
                disconnected.append(connection)
                
        # Remove disconnected clients
        for conn in disconnected:
            self.active_connections.remove(conn)
            
    def generate_dynamic_charts(self, assessment_data: dict) -> dict:
        """Generate all charts and return as HTML strings"""
        try:
            charts = {}
            
            # Generate all chart types
            risk_gauge = create_risk_gauge(assessment_data)
            constitution_chart = create_constitution_chart(assessment_data)
            health_radar = create_health_radar(assessment_data)
            cognitive_summary = create_cognitive_summary(assessment_data)
            compact_summary = create_compact_result_summary(assessment_data)
            
            # Convert to HTML
            charts['risk_gauge'] = pio.to_html(risk_gauge, include_plotlyjs='cdn', div_id='risk-gauge-chart')
            charts['constitution'] = pio.to_html(constitution_chart, include_plotlyjs='cdn', div_id='constitution-chart')
            charts['health_radar'] = pio.to_html(health_radar, include_plotlyjs='cdn', div_id='health-radar-chart')
            charts['cognitive'] = pio.to_html(cognitive_summary, include_plotlyjs='cdn', div_id='cognitive-chart')
            charts['compact'] = pio.to_html(compact_summary, include_plotlyjs='cdn', div_id='compact-chart')
            
            # Also return JSON data for dynamic updates
            charts['data'] = {
                'risk_gauge_data': risk_gauge.to_dict(),
                'constitution_data': constitution_chart.to_dict(),
                'health_radar_data': health_radar.to_dict(),
                'cognitive_data': cognitive_summary.to_dict(),
                'compact_data': compact_summary.to_dict()
            }
            
            return charts
            
        except Exception as e:
            print(f"Error generating charts: {e}")
            return {"error": str(e)}
            
    def get_dashboard_html(self) -> str:
        """Generate complete dashboard HTML with real-time features"""
        return '''
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>🏥⚕️ Dynamic Assessment Dashboard</title>
    <script src="https://cdn.plot.ly/plotly-latest.min.js"></script>
    <style>
        body {
            font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
            margin: 0;
            padding: 20px;
            background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
            min-height: 100vh;
        }
        .container {
            max-width: 1400px;
            margin: 0 auto;
            background: white;
            border-radius: 15px;
            padding: 30px;
            box-shadow: 0 20px 40px rgba(0,0,0,0.1);
        }
        .header {
            text-align: center;
            margin-bottom: 30px;
            padding: 20px;
            background: linear-gradient(90deg, #4CAF50, #2196F3);
            color: white;
            border-radius: 10px;
        }
        .controls {
            display: flex;
            gap: 20px;
            margin-bottom: 30px;
            flex-wrap: wrap;
            padding: 20px;
            background: #f8f9fa;
            border-radius: 10px;
        }
        .control-group {
            display: flex;
            flex-direction: column;
            gap: 10px;
        }
        .slider-container {
            display: flex;
            align-items: center;
            gap: 15px;
            margin-bottom: 10px;
        }
        .slider {
            width: 200px;
            margin: 0 10px;
        }
        .value-display {
            min-width: 40px;
            text-align: center;
            font-weight: bold;
            color: #333;
        }
        .chart-container {
            margin: 20px 0;
            border: 1px solid #ddd;
            border-radius: 10px;
            overflow: hidden;
        }
        .status {
            position: fixed;
            top: 20px;
            right: 20px;
            padding: 10px 20px;
            border-radius: 25px;
            color: white;
            font-weight: bold;
            z-index: 1000;
        }
        .status.connected { background: #4CAF50; }
        .status.disconnected { background: #f44336; }
        .status.updating { background: #ff9800; }
        button {
            padding: 10px 20px;
            border: none;
            border-radius: 5px;
            cursor: pointer;
            font-weight: bold;
            transition: all 0.3s ease;
        }
        .btn-primary { background: #2196F3; color: white; }
        .btn-success { background: #4CAF50; color: white; }
        .btn-warning { background: #ff9800; color: white; }
        .btn-danger { background: #f44336; color: white; }
        button:hover { transform: translateY(-2px); box-shadow: 0 4px 8px rgba(0,0,0,0.2); }
        .grid {
            display: grid;
            grid-template-columns: repeat(auto-fit, minmax(500px, 1fr));
            gap: 20px;
            margin-top: 20px;
        }
    </style>
</head>
<body>
    <div class="container">
        <div class="header">
            <h1>🏥⚕️ Dynamic Assessment Dashboard</h1>
            <p>Real-time Ayurvedic + Cognitive Health Visualization</p>
            <div id="status" class="status connected">🔄 Connected</div>
        </div>
        
        <div class="controls">
            <div class="control-group">
                <h3>🏛️ Ayurvedic Constitution</h3>
                <div class="slider-container">
                    <label>Vata Score:</label>
                    <input type="range" id="vata" class="slider" min="0" max="100" value="75">
                    <span id="vata-value" class="value-display">75</span>
                </div>
                <div class="slider-container">
                    <label>Pitta Score:</label>
                    <input type="range" id="pitta" class="slider" min="0" max="100" value="45">
                    <span id="pitta-value" class="value-display">45</span>
                </div>
                <div class="slider-container">
                    <label>Kapha Score:</label>
                    <input type="range" id="kapha" class="slider" min="0" max="100" value="30">
                    <span id="kapha-value" class="value-display">30</span>
                </div>
            </div>
            
            <div class="control-group">
                <h3>💪 Physical Health</h3>
                <div class="slider-container">
                    <label>Sleep Quality:</label>
                    <input type="range" id="sleep" class="slider" min="0" max="100" value="35">
                    <span id="sleep-value" class="value-display">35</span>
                </div>
                <div class="slider-container">
                    <label>Digestion:</label>
                    <input type="range" id="digestion" class="slider" min="0" max="100" value="60">
                    <span id="digestion-value" class="value-display">60</span>
                </div>
                <div class="slider-container">
                    <label>Energy Levels:</label>
                    <input type="range" id="energy" class="slider" min="0" max="100" value="40">
                    <span id="energy-value" class="value-display">40</span>
                </div>
                <div class="slider-container">
                    <label>Stress Level:</label>
                    <input type="range" id="stress" class="slider" min="0" max="100" value="80">
                    <span id="stress-value" class="value-display">80</span>
                </div>
            </div>
            
            <div class="control-group">
                <h3>🧠 Cognitive Health</h3>
                <div class="slider-container">
                    <label>Memory:</label>
                    <input type="range" id="memory" class="slider" min="0" max="100" value="70">
                    <span id="memory-value" class="value-display">70</span>
                </div>
                <div class="slider-container">
                    <label>Attention:</label>
                    <input type="range" id="attention" class="slider" min="0" max="100" value="45">
                    <span id="attention-value" class="value-display">45</span>
                </div>
                <div class="slider-container">
                    <label>Processing Speed:</label>
                    <input type="range" id="processing" class="slider" min="0" max="100" value="55">
                    <span id="processing-value" class="value-display">55</span>
                </div>
                <div class="slider-container">
                    <label>Executive Function:</label>
                    <input type="range" id="executive" class="slider" min="0" max="100" value="40">
                    <span id="executive-value" class="value-display">40</span>
                </div>
            </div>
            
            <div class="control-group">
                <h3>⚙️ Controls</h3>
                <button id="random-btn" class="btn-primary">🎲 Random Data</button>
                <button id="reset-btn" class="btn-warning">🔄 Reset</button>
                <button id="save-btn" class="btn-success">💾 Save Assessment</button>
                <label>
                    <input type="checkbox" id="auto-update" checked> Auto Update
                </label>
            </div>
        </div>
        
        <div class="grid" id="charts-grid">
            <div id="risk-gauge-chart" class="chart-container"></div>
            <div id="constitution-chart" class="chart-container"></div>
            <div id="health-radar-chart" class="chart-container"></div>
            <div id="cognitive-chart" class="chart-container"></div>
            <div id="compact-chart" class="chart-container"></div>
        </div>
    </div>

    <script>
        // WebSocket connection for real-time updates
        const ws = new WebSocket(`ws://${window.location.host}/ws`);
        const statusElement = document.getElementById('status');
        
        ws.onopen = () => {
            statusElement.textContent = '🔄 Connected';
            statusElement.className = 'status connected';
        };
        
        ws.onclose = () => {
            statusElement.textContent = '❌ Disconnected';
            statusElement.className = 'status disconnected';
        };
        
        ws.onmessage = (event) => {
            const message = JSON.parse(event.data);
            if (message.type === 'chart_update' || message.type === 'broadcast_update') {
                updateChartsFromData(message.charts);
            }
        };
        
        // Slider management
        const sliders = ['vata', 'pitta', 'kapha', 'sleep', 'digestion', 'energy', 'stress', 'memory', 'attention', 'processing', 'executive'];
        
        sliders.forEach(sliderId => {
            const slider = document.getElementById(sliderId);
            const valueDisplay = document.getElementById(sliderId + '-value');
            
            slider.addEventListener('input', () => {
                valueDisplay.textContent = slider.value;
                if (document.getElementById('auto-update').checked) {
                    updateCharts();
                }
            });
        });
        
        // Control buttons
        document.getElementById('random-btn').addEventListener('click', generateRandomData);
        document.getElementById('reset-btn').addEventListener('click', resetToDefault);
        document.getElementById('save-btn').addEventListener('click', saveAssessment);
        
        function generateRandomData() {
            sliders.forEach(sliderId => {
                const slider = document.getElementById(sliderId);
                const randomValue = Math.floor(Math.random() * 81) + 20; // 20-100
                slider.value = randomValue;
                document.getElementById(sliderId + '-value').textContent = randomValue;
            });
            updateCharts();
        }
        
        function resetToDefault() {
            const defaults = {vata: 75, pitta: 45, kapha: 30, sleep: 35, digestion: 60, energy: 40, stress: 80, memory: 70, attention: 45, processing: 55, executive: 40};
            Object.entries(defaults).forEach(([key, value]) => {
                document.getElementById(key).value = value;
                document.getElementById(key + '-value').textContent = value;
            });
            updateCharts();
        }
        
        function getCurrentData() {
            return {
                user_id: 'WEB_USER_' + Date.now(),
                assessment_date: new Date().toISOString(),
                vata_score: parseInt(document.getElementById('vata').value),
                pitta_score: parseInt(document.getElementById('pitta').value),
                kapha_score: parseInt(document.getElementById('kapha').value),
                sleep_quality: parseInt(document.getElementById('sleep').value),
                digestion_score: parseInt(document.getElementById('digestion').value),
                energy_levels: parseInt(document.getElementById('energy').value),
                stress_level: parseInt(document.getElementById('stress').value),
                memory_score: parseInt(document.getElementById('memory').value),
                attention_score: parseInt(document.getElementById('attention').value),
                processing_speed: parseInt(document.getElementById('processing').value),
                executive_function: parseInt(document.getElementById('executive').value)
            };
        }
        
        async function updateCharts() {
            statusElement.textContent = '⏳ Updating...';
            statusElement.className = 'status updating';
            
            try {
                const data = getCurrentData();
                const response = await fetch('/api/charts/generate', {
                    method: 'POST',
                    headers: {'Content-Type': 'application/json'},
                    body: JSON.stringify(data)
                });
                
                const result = await response.json();
                updateChartsFromData(result.charts);
                
                statusElement.textContent = '✅ Updated';
                statusElement.className = 'status connected';
                
            } catch (error) {
                console.error('Update error:', error);
                statusElement.textContent = '❌ Error';
                statusElement.className = 'status disconnected';
            }
        }
        
        function updateChartsFromData(charts) {
            if (charts) {
                Object.entries(charts).forEach(([chartType, chartHtml]) => {
                    if (chartType !== 'data' && chartHtml && typeof chartHtml === 'string') {
                        const container = document.getElementById(chartType + '-chart');
                        if (container) {
                            container.innerHTML = chartHtml;
                        }
                    }
                });
            }
        }
        
        async function saveAssessment() {
            try {
                const data = getCurrentData();
                const response = await fetch('/api/assessment/create', {
                    method: 'POST',
                    headers: {'Content-Type': 'application/json'},
                    body: JSON.stringify(data)
                });
                
                const result = await response.json();
                alert('Assessment saved successfully!');
                
            } catch (error) {
                console.error('Save error:', error);
                alert('Error saving assessment');
            }
        }
        
        // Initial chart load
        updateCharts();
    </script>
</body>
</html>
        '''
        
    def run_server(self, host="127.0.0.1", port=8000, reload=False):
        """Run the web server"""
        print(f"🚀 Starting Dynamic Dashboard Web Server...")
        print(f"🌐 Dashboard URL: http://{host}:{port}")
        print(f"📊 API Documentation: http://{host}:{port}/docs")
        print(f"🔄 WebSocket Endpoint: ws://{host}:{port}/ws")
        
        uvicorn.run(self.app, host=host, port=port, reload=reload)

# Create web dashboard manager
web_dashboard = WebDashboardManager()

print("🌐 Web Integration Ready!")
print("🚀 Complete FastAPI application with:")
print("   ✅ Real-time WebSocket updates")
print("   ✅ REST API endpoints")
print("   ✅ Interactive HTML dashboard")
print("   ✅ Data validation")
print("   ✅ Cross-origin support (CORS)")
print("   ✅ Auto-generated API docs")

print("\n🎯 To run the web server:")
print("   web_dashboard.run_server()")
print("   Then visit: http://127.0.0.1:8000")
```
